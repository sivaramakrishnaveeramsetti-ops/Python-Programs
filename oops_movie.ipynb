```python
class Movie:
  "list of movies in 2024"
  def __init__(self,title,hero,herione):
    self.title=title
    self.hero=hero
    self.herione=herione

  def info(self):
    print("Movie Name :",self.title)
    print("Hero Name: ",self.hero)
    print("Herione Name: ",self.herione)


list_of_movies=[]
while True:
  title=input("enter movie title: ")
  hero=input("enter hero name: ")
  herione=input("enter herione name : ")
  m=Movie(title,hero,herione)
  list_of_movies.append(m)
  print("Movie added successfully.....")
  option=input("do you want to add another movie {yes/no}")
  if option.lower()=="no":
    print("Total movies:", len(list_of_movies))
    break
print("----ALL MOVIES INFORMATION----")
for movie in list_of_movies:
    movie.info()
    print()

```
